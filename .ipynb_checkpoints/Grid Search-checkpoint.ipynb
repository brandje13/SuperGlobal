```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style for the notebook
sns.set_theme(style="whitegrid")

# Load the dataset
df = pd.read_csv('grid_search_ILIAS.csv')

# To avoid plotting the exact same mAP points hundreds of thousands of times, 
# we'll extract the unique mapping of backbone + M value to mAP.
global_df = df[['global_bb', 'global_m', 'global_map']].drop_duplicates()
local_df = df[['local_bb', 'local_m', 'local_map']].drop_duplicates()
sem_df = df[['sem_bb', 'sem_map']].drop_duplicates()
```

```python
# 1. Clean up long backbone names for better X-axis readability
# Removes file paths, extensions, and repository tags
df['global_bb_clean'] = df['global_bb'].str.replace(r'.\weights\CVPR2022_', '', regex=False).str.replace(r'.pyth', '', regex=False)
df['local_bb_clean'] = df['local_bb'].str.replace(r'_dinov2.lvd142m', '', regex=False).str.replace(r'vit_', '', regex=False)
df['sem_bb_clean'] = df['sem_bb'].str.split('/').str[-1] 

# 2. Extract mappings and force M values to be strings (Categorical)
global_df = df[['global_bb_clean', 'global_m', 'global_map']].drop_duplicates()
global_df['global_m'] = global_df['global_m'].astype(str) # Fixes the legend hue numbers

local_df = df[['local_bb_clean', 'local_m', 'local_map']].drop_duplicates()
local_df['local_m'] = local_df['local_m'].astype(str) # Fixes the legend hue numbers

sem_df = df[['sem_bb_clean', 'sem_map']].drop_duplicates()

# 3. Plotting
fig, axes = plt.subplots(3, 1, figsize=(14, 18))

# Global Plot
sns.barplot(data=global_df, x='global_bb_clean', y='global_map', hue='global_m', palette='viridis', ax=axes[0])
axes[0].set_title('Global Backbone mAP across Rerank values (M)', fontsize=14)
axes[0].set_xticklabels(axes[0].get_xticklabels(), rotation=20, ha='right')
axes[0].set_ylabel('mAP')
axes[0].set_xlabel('')
axes[0].legend(title='Global M', bbox_to_anchor=(1.01, 1), loc='upper left')

# Local Plot
sns.barplot(data=local_df, x='local_bb_clean', y='local_map', hue='local_m', palette='magma', ax=axes[1])
axes[1].set_title('Local Backbone mAP across Rerank values (M)', fontsize=14)
axes[1].set_xticklabels(axes[1].get_xticklabels(), rotation=20, ha='right')
axes[1].set_ylabel('mAP')
axes[1].set_xlabel('')
axes[1].legend(title='Local M', bbox_to_anchor=(1.01, 1), loc='upper left')

# Semantic Plot
sns.barplot(data=sem_df, x='sem_bb_clean', y='sem_map', palette='coolwarm', ax=axes[2])
axes[2].set_title('Semantic Backbone mAP', fontsize=14)
axes[2].set_xticklabels(axes[2].get_xticklabels(), rotation=20, ha='right')
axes[2].set_ylabel('mAP')
axes[2].set_xlabel('')

plt.tight_layout()
plt.show()
```

```python
# Get the top 5 configurations based on F3 score
top_configs = df.nlargest(5, 'f3').copy()

# Create a simplified label for the X-axis (e.g., "Rank 1: Config 1245")
top_configs['config_label'] = [f"Rank {i+1}\n(Index: {idx})" for i, idx in enumerate(top_configs.index)]

# Set up the plot
sns.set_theme(style="whitegrid")
fig, ax = plt.subplots(figsize=(10, 6))

# Plot the stacked bars
top_configs.set_index('config_label')[['global_time', 'local_time', 'sem_time']].plot(
    kind='bar', 
    stacked=True, 
    color=['#440154', '#21918c', '#fde725'], # Viridis color palette
    ax=ax,
    width=0.6
)

plt.title('Compute Time Breakdown for Top 5 Configurations', fontsize=15, fontweight='bold')
plt.ylabel('Time (Seconds)', fontsize=12)
plt.xlabel('Configuration', fontsize=12)
plt.xticks(rotation=0)

# Clean up the legend
plt.legend(['Global Time (CVNet)', 'Local Time (DINOv2)', 'Semantic Time (SigLIP)'], 
           loc='upper right', bbox_to_anchor=(1.35, 1))

plt.tight_layout()
plt.show()
```

```python
# 1. Filter for your absolute best meta-configuration
ablation_df = df[
    (df['global_bb'].str.contains('CVNet_R101')) &
    (df['local_bb'].str.contains('giant_patch14_reg4_dinov2.lvd142m_224')) &
    (df['sem_bb'].str.contains('siglip-large-patch16-384')) &
    (df['mode'] == 'majority') &
    (df['top_k'] == 50)
].copy()

# 2. Sort by local_m (from highest M to 0) to show the decay
ablation_df = ablation_df.sort_values('local_m', ascending=False)

# 3. Create the plot with twin Y-axes
sns.set_theme(style="whitegrid")
fig, ax1 = plt.subplots(figsize=(10, 6))

# Plot Total Time on the primary Y-axis (left)
color1 = '#d9534f' # Red
ax1.plot(ablation_df['local_m'].astype(str), ablation_df['total_time'], color=color1, marker='o', linewidth=2.5, markersize=8)
ax1.set_xlabel('Local M (Rerank Value)', fontsize=12, fontweight='bold')
ax1.set_ylabel('Total Compute Time (Seconds)', color=color1, fontsize=12, fontweight='bold')
ax1.tick_params(axis='y', labelcolor=color1)
ax1.invert_xaxis() # Read left-to-right as M decreases

# Create a secondary Y-axis (right) for the F3 Score
ax2 = ax1.twinx()  
color2 = '#5bc0de' # Blue
ax2.plot(ablation_df['local_m'].astype(str), ablation_df['f3'], color=color2, marker='s', linewidth=2.5, markersize=8)
ax2.set_ylabel('F3 Score', color=color2, fontsize=12, fontweight='bold')
ax2.tick_params(axis='y', labelcolor=color2)

plt.title('Ablation Study: Compute Time vs. Accuracy Decay (Decreasing M)', fontsize=14, pad=15)
fig.tight_layout()
plt.show()
```

```python
import plotly.express as px

# 1. Clean up strings for the plot labels
df['Global'] = df['global_bb'].str.replace(r'.\weights\CVPR2022_', '', regex=False).str.replace(r'.pyth', '', regex=False)
df['Local'] = df['local_bb'].str.replace(r'_dinov2.lvd142m', '', regex=False)
df['Semantic'] = df['sem_bb'].str.split('/').str[-1]
df['Top-K'] = df['top_k'].astype(str)
df['Mode'] = df['mode'].str.capitalize()

# 2. Sample the data so the browser doesn't crash (take Top 500 and a random 2000)
top_500 = df.nlargest(100, 'f3')
random_sample = df.sample(200)
plot_df = pd.concat([top_500, random_sample]).drop_duplicates()

# 3. Create the Interactive Parallel Categories plot
fig = px.parallel_categories(
    plot_df,
    dimensions=['Global', 'Local', 'Semantic', 'Top-K', 'Mode'],
    color="f3",
    color_continuous_scale=px.colors.sequential.Inferno,
    title="Hyperparameter Flow: Which configurations yield the highest F3? (Interactive)"
)

# 4. Tweak visual aesthetics
fig.update_layout(
    margin=dict(l=50, r=50, b=50, t=80),
    font=dict(size=11)
)

# Display the interactive plot in the notebook
fig.show()
```
