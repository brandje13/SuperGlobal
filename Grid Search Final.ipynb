```python
# 1. Clean up long backbone names for better X-axis readability
# Removes file paths, extensions, and repository tags
df['global_bb_clean'] = df['global_bb'].str.replace(r'.\weights\CVPR2022_', '', regex=False).str.replace(r'.pyth', '', regex=False)
df['local_bb_clean'] = df['local_bb'].str.replace(r'_dinov2.lvd142m', '', regex=False).str.replace(r'vit_', '', regex=False)
df['sem_bb_clean'] = df['sem_bb'].str.split('/').str[-1] 

# 2. Extract mappings and force M values to be strings (Categorical)
global_df = df[['global_bb_clean', 'global_m', 'global_map']].drop_duplicates()
global_df['global_m'] = global_df['global_m'].astype(str) # Fixes the legend hue numbers

local_df = df[['local_bb_clean', 'local_m', 'local_map']].drop_duplicates()
local_df['local_m'] = local_df['local_m'].astype(str) # Fixes the legend hue numbers

sem_df = df[['sem_bb_clean', 'sem_map']].drop_duplicates()

# 3. Plotting
fig, axes = plt.subplots(3, 1, figsize=(14, 18))

# Global Plot
sns.barplot(data=global_df, x='global_bb_clean', y='global_map', hue='global_m', palette='viridis', ax=axes[0])
axes[0].set_title('Global Backbone mAP across Rerank values (M)', fontsize=14)
axes[0].set_xticklabels(axes[0].get_xticklabels(), rotation=20, ha='right')
axes[0].set_ylabel('mAP')
axes[0].set_xlabel('')
axes[0].legend(title='Global M', bbox_to_anchor=(1.01, 1), loc='upper left')

# Local Plot
sns.barplot(data=local_df, x='local_bb_clean', y='local_map', hue='local_m', palette='magma', ax=axes[1])
axes[1].set_title('Local Backbone mAP across Rerank values (M)', fontsize=14)
axes[1].set_xticklabels(axes[1].get_xticklabels(), rotation=20, ha='right')
axes[1].set_ylabel('mAP')
axes[1].set_xlabel('')
axes[1].legend(title='Local M', bbox_to_anchor=(1.01, 1), loc='upper left')

# Semantic Plot
sns.barplot(data=sem_df, x='sem_bb_clean', y='sem_map', palette='coolwarm', ax=axes[2])
axes[2].set_title('Semantic Backbone mAP', fontsize=14)
axes[2].set_xticklabels(axes[2].get_xticklabels(), rotation=20, ha='right')
axes[2].set_ylabel('mAP')
axes[2].set_xlabel('')

plt.tight_layout()
plt.show()
```

```python
import plotly.express as px

# 1. Clean up strings for the plot labels
df['Global'] = df['global_bb'].str.replace(r'.\weights\CVPR2022_', '', regex=False).str.replace(r'.pyth', '', regex=False)
df['Local'] = df['local_bb'].str.replace(r'_dinov2.lvd142m', '', regex=False)
df['Semantic'] = df['sem_bb'].str.split('/').str[-1]
df['Top-K'] = df['top_k'].astype(str)
df['Mode'] = df['mode'].str.capitalize()

# 2. Sample the data so the browser doesn't crash (take Top 500 and a random 2000)
top_500 = df.nlargest(100, 'f3')
random_sample = df.sample(200)
plot_df = pd.concat([top_500, random_sample]).drop_duplicates()

# 3. Create the Interactive Parallel Categories plot
fig = px.parallel_categories(
    plot_df,
    dimensions=['Global', 'Local', 'Semantic', 'Top-K', 'Mode'],
    color="f3",
    color_continuous_scale=px.colors.sequential.Inferno,
    title="Hyperparameter Flow: Which configurations yield the highest F3? (Interactive)"
)

# 4. Tweak visual aesthetics
fig.update_layout(
    margin=dict(l=50, r=50, b=50, t=80),
    font=dict(size=11)
)

# Display the interactive plot in the notebook
fig.show()
```

```python
# CELL: CALCULATE THE PARETO OPTIMAL TRADE-OFF (UTOPIA DISTANCE)
import numpy as np

# 1. Normalize F3 and Time to a 0.0 - 1.0 scale
df['f3_norm'] = (df['f3'] - df['f3'].min()) / (df['f3'].max() - df['f3'].min())
df['time_norm'] = (df['total_time'] - df['total_time'].min()) / (df['total_time'].max() - df['total_time'].min())

# 2. Calculate Euclidean Distance to the theoretical "Utopia" point (Time=0, F3=1)
df['utopia_distance'] = np.sqrt((1 - df['f3_norm'])**2 + (df['time_norm'])**2)

# 3. Extract the top 10 best configurations based on Pareto distance
top_10_tradeoffs = df.nsmallest(10, 'utopia_distance')

# 4. Select only the human-readable columns for the thesis table
columns_to_show = [
    'Mode', 
    'Top-K', 
    'Global', 
    'global_m',
    'Local', 
    'local_m',  
    'Semantic', 
    'precision', 
    'recall', 
    'f3', 
    'total_time', 
    'utopia_distance'
]

# Create a clean subset
clean_table = top_10_tradeoffs[columns_to_show].copy()

# Format the numbers so the table looks highly professional
clean_table['precision'] = clean_table['precision'].round(4)
clean_table['recall'] = clean_table['recall'].round(4)
clean_table['f3'] = clean_table['f3'].round(4)
clean_table['total_time'] = clean_table['total_time'].round(1).astype(str) + "s"
clean_table['utopia_distance'] = clean_table['utopia_distance'].round(4)

# Optional: Rename the columns so they look better in the final screenshot
clean_table = clean_table.rename(columns={
    'global_m': 'Global M',
    'local_m': 'Local M',
    'precision': 'Precision',
    'recall': 'Recall',
    'f3': 'F3 Score',
    'total_time': 'Total Time',
    'utopia_distance': 'Utopia Dist.'
})

print("--- TOP 10 PARETO OPTIMAL CONFIGURATIONS ---")
display(clean_table)
```
